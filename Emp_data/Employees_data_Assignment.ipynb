### Step 2: Data Preprocessing



```python
# Check Data Types
df.info()
```

```python
# Check Missing Values
df.isnull().sum()
```

```python
# Check Duplicate Records
df.duplicated().sum()
```

```python
# Standardize Gender Values
df['Gender'] = df['Gender'].str.title()
```

```python
# Standardize Location Names
df['Location'] = df['Location'].str.title()
```

### Step 3: Data Analysis


```python
# Profiling
df.describe()
```

```python
# Salary Distribution
plt.hist(df['Salary'])

plt.xlabel("Salary")

plt.ylabel("Employees")

plt.title("Salary Distribution")

plt.show()
```

```python
# Age Distribution

plt.hist(df['Age'])

plt.xlabel("Age")

plt.ylabel("Employees")

plt.title("Age Distribution")

plt.show()
```

```python
# Gender Distribution
df['Gender'].value_counts()
```

```python
# Graph
df['Gender'].value_counts().plot(kind='bar')

plt.show()
```

```python
# Average Salary by Job Title

df.groupby('JobTitle')['Salary'].mean()
```

```python
# Average Salary by Gender
df.groupby('Gender')['Salary'].mean()
```

```python
# Average Salary by Location
df.groupby('Location')['Salary'].mean()
```

```python
# Highest Paid Employees
df.sort_values(by='Salary', ascending=False).head(10)
```

```python
# Lowest Paid Employees
df.sort_values(by='Salary').head(10)
```

```python
# Salary by Job Title (Visualization)
df.groupby('JobTitle')['Salary'].mean().plot(kind='bar')

plt.xticks(rotation=30)

plt.ylabel("Average Salary")

plt.show()
```

```python
# Salary by Gender
df.groupby('Gender')['Salary'].mean().plot(kind='bar')

plt.ylabel("Average Salary")

plt.show()
```

```python
# Salary by Location
df.groupby('Location')['Salary'].mean().plot(kind='bar')

plt.ylabel("Average Salary")

plt.show()
```

```python
#Correlation Analysis
df[['Age','Salary']].corr()
```

### Step 4: Insights & Recommendations

Insight 1

Observation:

Employees working in New York and San Francisco receive higher average salaries than other locations.

Recommendation:

Maintain location-based salary adjustments to stay competitive in expensive cities.

Insight 2

Observation:

Some job titles (such as Software Engineer or Data Scientist) have higher average salaries.

Recommendation:

Review salary structures regularly to remain competitive for high-demand technical roles.

Insight 3

Observation:

If male and female employees in the same job role have different average salaries, it may indicate a gender pay gap.

Recommendation:

Conduct periodic pay-equity reviews and ensure equal pay for equal work.

Insight 4

Observation:

Employees with more experience (higher age) generally tend to earn higher salaries.

Recommendation:

Create transparent promotion and salary growth policies based on performance and experience.

Insight 5

Observation:

A few employees receive significantly higher salaries than the majority.

Recommendation:

Review salary outliers to ensure they are justified by role, performance, or experience.
